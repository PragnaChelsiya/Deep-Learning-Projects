```python
# TASK 1: Actual Class Labels and Predictions

import numpy as np

# Actual class labels
y_actual = np.array([1, 0, 1, 1, 0])

# Predicted outputs
y_predicted = np.array([0.9, 0.2, 0.8, 0.7, 0.1])

print("Actual Labels:", y_actual)
print("Predicted Outputs:", y_predicted)
```

```python
# TASK 2: Calculate MSE and Cross-Entropy Loss

# Mean Squared Error
mse = np.mean((y_actual - y_predicted) ** 2)

# Cross-Entropy Loss
epsilon = 1e-15
y_predicted = np.clip(y_predicted, epsilon, 1 - epsilon)

cross_entropy = -np.mean(
    y_actual * np.log(y_predicted) +
    (1 - y_actual) * np.log(1 - y_predicted)
)

print("Mean Squared Error:", mse)
print("Cross-Entropy Loss:", cross_entropy)
```

```python
# TASK 3: Compare the Results

print("\nComparison:")
print("MSE is mainly used for regression problems.")
print("Cross-Entropy Loss is mainly used for classification problems.")
print("Cross-Entropy Loss is commonly preferred for classification.")
```
